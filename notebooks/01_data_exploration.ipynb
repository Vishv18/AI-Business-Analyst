```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

orders[date_columns].head()
```

```python
for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

orders[date_columns].dtypes
```

```python
orders[date_columns].isnull().sum()
```

```python
orders["order_status"].value_counts()
```

```python
orders.groupby("order_status")[
    [
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date"
    ]
].apply(lambda group: group.isnull().sum())
```

```python
delivered_missing_dates = orders[
    (orders["order_status"] == "delivered") &
    (
        orders["order_approved_at"].isnull() |
        orders["order_delivered_carrier_date"].isnull() |
        orders["order_delivered_customer_date"].isnull()
    )
]

delivered_missing_dates[
    [
        "order_id",
        "order_status",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
]
```

```python
delivered_orders = orders[
    orders["order_status"] == "delivered"
]

print("Total delivered orders:", len(delivered_orders))

print(
    "Missing approval timestamps:",
    delivered_orders["order_approved_at"].isnull().sum()
)

print(
    "Missing carrier timestamps:",
    delivered_orders["order_delivered_carrier_date"].isnull().sum()
)

print(
    "Missing customer delivery timestamps:",
    delivered_orders["order_delivered_customer_date"].isnull().sum()
)
```

```python
print(
    "First order purchase date:",
    orders["order_purchase_timestamp"].min()
)

print(
    "Last order purchase date:",
    orders["order_purchase_timestamp"].max()
)
```

```python
orders["purchase_month"] = (
    orders["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_orders = (
    orders.groupby("purchase_month")["order_id"]
    .count()
    .reset_index(name="total_orders")
)

monthly_orders
```

```python
print("Earliest purchase:", orders["order_purchase_timestamp"].min())
print("Latest purchase:", orders["order_purchase_timestamp"].max())
```

```python
orders.groupby(
    orders["order_purchase_timestamp"].dt.to_period("M")
).agg(
    first_order_date=("order_purchase_timestamp", "min"),
    last_order_date=("order_purchase_timestamp", "max"),
    total_orders=("order_id", "count")
)
```
